```python

# 平成24年度・平成30年度の構造を確認

for year in ["24", "30"]:

    # ファイル名から対象ファイルを探す
    target = [
        f for f in files
        if os.path.basename(f).startswith(year + "ekibetsuhouchi")
    ]

    if len(target) == 0:
        print(year, "年度のファイルがありません")
        continue

    # Excelを読み込む
    df_check = pd.read_excel(
        target[0],
        sheet_name=0,
        header=None
    )

    print("平成" + year + "年度")
    display(df_check.iloc[:6, :14])
```

```python

# 14年間のデータを格納するリスト
all_data = []

# 平成24～30年度
for year in range(24, 31):

    # ファイル名を探す
    target = [
        f for f in files
        if os.path.basename(f).startswith(str(year) + "ekibet")
    ]

    if len(target) != 1:
        print("ファイルを確認してください：平成", year)
        continue

    # Excelを読み込む
    data = pd.read_excel(target[0], header=None)

    # 駅名と放置自転車台数を抽出
    data = data.iloc[:, [2, 3]].copy()
    data.columns = ["駅名", "放置自転車台数"]

    # 西暦年度を追加
    data["年度"] = year + 1988

    # リストに追加
    all_data.append(data)

# 令和元～7年度
for year in range(1, 8):

    target = [
        f for f in files
        if os.path.basename(f).startswith(f"{year:02d}ekibet")
    ]

    if len(target) != 1:
        print("ファイルを確認してください：令和", year)
        continue

    data = pd.read_excel(target[0], header=None)

    data = data.iloc[:, [2, 3]].copy()
    data.columns = ["駅名", "放置自転車台数"]

    data["年度"] = year + 2018

    all_data.append(data)

# 14年間のデータを縦に結合
df_all = pd.concat(all_data, ignore_index=True)

# データを確認
print("統合後の行数：", len(df_all))
display(df_all.head(10))
```

```python

# 年度と駅名で並べ替える
df_ml = df_station.sort_values(
    ["駅名", "年度"]
).copy()

# 同じ駅の翌年度の放置台数を取得
df_ml["翌年度の放置台数"] = (
    df_ml.groupby("駅名")["放置自転車台数"].shift(-1)
)

# 次に記録されている年度を取得
df_ml["次の記録年度"] = (
    df_ml.groupby("駅名")["年度"].shift(-1)
)

# 翌年度の記録がある行だけを残す
df_ml = df_ml[
    df_ml["次の記録年度"] == df_ml["年度"] + 1
].copy()

# 翌年度の放置台数が100台以上なら1、それ以外は0
df_ml["翌年度リスク"] = (
    df_ml["翌年度の放置台数"] >= 100
).astype(int)

# 結果を確認
print("機械学習に使えるデータ数：", len(df_ml))

display(df_ml.head(10))
```

```python

# 翌年度リスクが0と1の件数を確認する
print("【リスク別の件数】")
print(df_ml["翌年度リスク"].value_counts())

# 割合（%）を確認する
print("\n【リスク別の割合】")
print(
    (df_ml["翌年度リスク"].value_counts(normalize=True) * 100).round(1)
)

# 100台以上のリスクがある駅の割合を確認
risk_rate = df_ml["翌年度リスク"].mean() * 100

print(f"\nリスクありの割合：{risk_rate:.1f}%")
```

```python

# 学習・検証・テスト用に分割する

# 2022年度までの情報から翌年度を予測するデータ
train = df_ml[df_ml["年度"] <= 2022].copy()

# 2023年度の情報から2024年度を予測
val = df_ml[df_ml["年度"] == 2023].copy()

# 2024年度の情報から2025年度を予測
test = df_ml[df_ml["年度"] == 2024].copy()

# 件数を確認
print("学習用：", len(train), "件")
print("検証用：", len(val), "件")
print("テスト用：", len(test), "件")

# 対象年度の確認
print("学習用の入力年度：", train["年度"].min(), "〜", train["年度"].max())
print("検証用の入力年度：", val["年度"].unique())
print("テスト用の入力年度：", test["年度"].unique())
```

```python

# ランダムフォレストを読み込む
from sklearn.ensemble import RandomForestClassifier

# AIに入力する情報（説明変数）
X_train = train[["放置自転車台数"]]

# AIが学習する正解（目的変数）
y_train = train["翌年度リスク"]

# ランダムフォレストのモデルを作る
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42,
    class_weight="balanced"
)

# AIに学習させる
model.fit(X_train, y_train)

print("AIの学習が完了しました！")
```

```python

# AIの予測精度を評価する機能を読み込む
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix

# 検証用データの入力情報
X_val = val[["放置自転車台数"]]

# 検証用データの正解
y_val = val["翌年度リスク"]

# 学習済みAIで2024年度のリスクを予測
y_pred = model.predict(X_val)

# 予測精度を計算する
accuracy = accuracy_score(y_val, y_pred)
precision = precision_score(y_val, y_pred, zero_division=0)
recall = recall_score(y_val, y_pred, zero_division=0)
f1 = f1_score(y_val, y_pred, zero_division=0)

# 結果を表示する
print(f"正解率（Accuracy）：{accuracy:.3f}")
print(f"適合率（Precision）：{precision:.3f}")
print(f"再現率（Recall）：{recall:.3f}")
print(f"F1スコア：{f1:.3f}")

# 混同行列を表示する
print("\n混同行列")
print(confusion_matrix(y_val, y_pred, labels=[0, 1]))
```

```python

# 単純な予測方法：
# 今年100台以上なら、翌年度もリスクありと予測

simple_pred = (
    X_val["放置自転車台数"] >= 100
).astype(int)

# AIと単純な方法の成績を比較
print("【AIモデル】")
print("正解率：", round(accuracy_score(y_val, y_pred), 3))
print("再現率：", round(recall_score(y_val, y_pred), 3))
print("F1：", round(f1_score(y_val, y_pred), 3))

print("\n【単純な100台判定】")
print("正解率：", round(accuracy_score(y_val, simple_pred), 3))
print("再現率：", round(recall_score(y_val, simple_pred), 3))
print("F1：", round(f1_score(y_val, simple_pred), 3))
```

```python

# 年度と駅名の順番に並べる
df_ml2 = df_station.sort_values(
    ["駅名", "年度"]
).copy()

# 前年の放置自転車台数を取得する
df_ml2["前年の放置台数"] = (
    df_ml2.groupby("駅名")["放置自転車台数"].shift(1)
)

# 前年の年度を取得する
df_ml2["前年の年度"] = (
    df_ml2.groupby("駅名")["年度"].shift(1)
)

# 前年から何台増減したか計算する
df_ml2["前年差"] = (
    df_ml2["放置自転車台数"] -
    df_ml2["前年の放置台数"]
)

# 実際に連続した年度のデータだけ残す
df_ml2 = df_ml2[
    df_ml2["前年の年度"] == df_ml2["年度"] - 1
].copy()

# 結果を表示する
display(
    df_ml2[[
        "年度",
        "駅名",
        "前年の放置台数",
        "放置自転車台数",
        "前年差"
    ]].head(10)
)
```

```python

# 同じ駅の翌年度の放置台数を取得する
df_ml2["翌年度の放置台数"] = (
    df_ml2.groupby("駅名")["放置自転車台数"].shift(-1)
)

# 次の記録年度を取得する
df_ml2["次の記録年度"] = (
    df_ml2.groupby("駅名")["年度"].shift(-1)
)

# 翌年度の記録があるデータだけ残す
df_ml2 = df_ml2[
    df_ml2["次の記録年度"] == df_ml2["年度"] + 1
].copy()

# 翌年度100台以上なら1、それ以外なら0
df_ml2["翌年度リスク"] = (
    df_ml2["翌年度の放置台数"] >= 100
).astype(int)

# データ件数を確認
print("改良モデル用データ件数：", len(df_ml2))

# 最初の10行を表示
display(df_ml2.head(10))
```

```python

# 3年間が連続しているデータだけを残す
df_ml2 = df_ml2[
    (df_ml2["前年の年度"] == df_ml2["年度"] - 1) &
    (df_ml2["次の記録年度"] == df_ml2["年度"] + 1)
].copy()

# AIに入力する3つの特徴量
features = [
    "放置自転車台数",
    "前年の放置台数",
    "前年差"
]

# 学習用：2023年度までの結果を学習
train2 = df_ml2[df_ml2["年度"] <= 2022].copy()

# 検証用：2024年度の結果を予測
val2 = df_ml2[df_ml2["年度"] == 2023].copy()

# テスト用：2025年度の結果を予測
test2 = df_ml2[df_ml2["年度"] == 2024].copy()

# 件数を表示
print("学習用：", len(train2), "件")
print("検証用：", len(val2), "件")
print("テスト用：", len(test2), "件")

# AIに入力するデータを確認
display(train2[features].head())
```

```python

# ランダムフォレストを読み込む
from sklearn.ensemble import RandomForestClassifier

# 学習に使う3つの特徴量を指定する
X_train2 = train2[features]

# 正解データを指定する
y_train2 = train2["翌年度リスク"]

# 改良版のAIモデルを作成する
model2 = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42,
    class_weight="balanced"
)

# AIに学習させる
model2.fit(X_train2, y_train2)

# 学習完了を確認する
print("改良モデルの学習が完了しました！")

# AIが重視した特徴量を表示する
importance = pd.DataFrame({
    "特徴量": features,
    "重要度": model2.feature_importances_
})

display(importance.sort_values("重要度", ascending=False))
```

```python

# 台数予測に使うライブラリを読み込む
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# 過去の駅別放置自転車台数を読み込む
history = pd.read_csv("station_history.csv")

# データの先頭5行を確認する
print(history.head())
```

```python

# 年度と駅名の順に並べる
history = history.sort_values(["駅名", "年度"]).copy()

# 同じ駅の翌年度の放置自転車台数を取得
history["翌年度の台数"] = history.groupby("駅名")["放置自転車台数"].shift(-1)

# 翌年度の年度も取得する
history["翌年度"] = history.groupby("駅名")["年度"].shift(-1)

# 年度が連続しているデータだけを残す
train_data = history[
    (history["翌年度"] == history["年度"] + 1)
    & history["翌年度の台数"].notna()
].copy()

# 結果を確認
print(train_data[["年度", "駅名", "放置自転車台数", "翌年度の台数"]].head(10))
print("学習に利用できるデータ数：", len(train_data))
```

```python

# 2024年度より前のデータを学習用にする
# 例：2022年度 → 2023年度、2023年度 → 2024年度
train = train_data[train_data["年度"] < 2024].copy()

# 2024年度 → 2025年度のデータをテスト用にする
test = train_data[train_data["年度"] == 2024].copy()

# AIに入力するデータ（前年の放置自転車台数）
X_train = train[["放置自転車台数"]]
X_test = test[["放置自転車台数"]]

# AIに予測させたい正解データ（翌年度の放置自転車台数）
y_train = train["翌年度の台数"]
y_test = test["翌年度の台数"]

# データ件数を確認
print("学習データ：", len(X_train), "件")
print("テストデータ：", len(X_test), "件")
```

```python

# ランダムフォレスト回帰モデルを読み込む
from sklearn.ensemble import RandomForestRegressor

# 予測精度を評価するためのライブラリ
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# AIモデルを作成する
model_reg = RandomForestRegressor(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

# 過去のデータを使って学習する
model_reg.fit(X_train, y_train)

# 2024年度の台数から2025年度の台数を予測する
y_pred = model_reg.predict(X_test)

# 予測台数がマイナスにならないようにする
y_pred = np.maximum(y_pred, 0)

# MAE：平均すると何台ずれているか
mae = mean_absolute_error(y_test, y_pred)

# RMSE：大きな予測ミスを重く評価する指標
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# 比較対象：2025年度も2024年度と同じ台数だと予測
baseline_pred = X_test["放置自転車台数"].to_numpy()

baseline_mae = mean_absolute_error(y_test, baseline_pred)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_pred))

# 結果を表示
print("【AIモデル】")
print(f"MAE：{mae:.2f} 台")
print(f"RMSE：{rmse:.2f} 台")

print("\n【前年と同じ台数と予測した場合】")
print(f"MAE：{baseline_mae:.2f} 台")
print(f"RMSE：{baseline_rmse:.2f} 台")

# 実際の予測結果を10件表示
results = test[["駅名", "放置自転車台数", "翌年度の台数"]].copy()
results["2025年度のAI予測台数"] = np.rint(y_pred).astype(int)

print("\n【予測結果のサンプル】")
print(results.head(10))
```

```python

# 学習済みモデルを保存するためのライブラリ
import joblib
from sklearn.ensemble import RandomForestRegressor

# 2025年度までのデータを学習に使用する
X_all = train_data[["放置自転車台数"]]
y_all = train_data["翌年度の台数"]

# 2026年度の予測に使うモデルを作成
final_reg_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

# すべての学習データで学習し直す
final_reg_model.fit(X_all, y_all)

# モデルを保存
joblib.dump(final_reg_model, "bicycle_count_model.pkl")

print("2026年度予測用モデルを保存しました")
print("学習データ数：", len(X_all))
```

```python

# 2025年度の駅別データを読み込む
stations_2025 = pd.read_csv("station_data_2025.csv")

# 2025年度の放置自転車台数を入力データにする
X_2025 = stations_2025[["放置自転車台数"]]

# 学習済み回帰モデルで2026年度の台数を予測する
pred_2026 = final_reg_model.predict(X_2025)

# マイナスの予測値を0に補正し、整数にする
stations_2025["2026年度予測台数"] = (
    np.rint(np.maximum(pred_2026, 0)).astype(int)
)

# 2025年度からの予測増減台数を計算する
stations_2025["予測増減台数"] = (
    stations_2025["2026年度予測台数"]
    - stations_2025["放置自転車台数"]
)

# 2025年度の放置台数が多い駅から20件を表示する
display(
    stations_2025[
        ["駅名", "放置自転車台数", "2026年度予測台数", "予測増減台数"]
    ]
    .sort_values("放置自転車台数", ascending=False)
    .head(20)
)
```
